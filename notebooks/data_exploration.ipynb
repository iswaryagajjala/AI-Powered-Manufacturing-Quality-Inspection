# NEU-CLS Dataset Exploration

Run this after `python src/preprocessing.py` has populated `data/processed/`.

This notebook inspects: class balance, sample images per class, and basic image statistics (size, aspect ratio) to confirm the preprocessing pipeline (resize/normalize) is appropriate before training.

```python
import sys
sys.path.append('../src')
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image
from dataset import discover_classes

DATA_DIR = Path('../data/processed')
classes = discover_classes(DATA_DIR, split='train')
print('Classes:', classes)
```

```python
# Class balance across splits
for split in ('train', 'val', 'test'):
    counts = {c: len(list((DATA_DIR / split / c).glob('*'))) for c in classes}
    print(split, counts)
```

```python
# Sample images, one per class
fig, axes = plt.subplots(1, len(classes), figsize=(4 * len(classes), 4))
for ax, cls in zip(axes, classes):
    sample_path = next((DATA_DIR / 'train' / cls).glob('*'))
    img = Image.open(sample_path)
    ax.imshow(img)
    ax.set_title(cls)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
